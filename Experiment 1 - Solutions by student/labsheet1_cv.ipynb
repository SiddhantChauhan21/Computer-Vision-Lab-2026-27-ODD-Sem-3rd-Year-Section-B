```python

import cv2
import os
import matplotlib.pyplot as plt
from google.colab import files
from IPython.display import display, Image
# 1. Create Output Folder

os.makedirs("outputs", exist_ok=True)
# 2. Upload Image

print("Please upload your image:")
uploaded = files.upload()

# Get uploaded filename
image_path = next(iter(uploaded))

print("\nImage selected:", image_path)
# 3. Load Image

img = cv2.imread(image_path)

if img is None:
    print("Error: Image not found!")
else:
    print("Image loaded successfully!")

# 4. Convert BGR to RGB

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
# 5. Image Properties

height, width, channels = img.shape

print("IMAGE PROPERTIES")

print("Width       :", width, "pixels")
print("Height      :", height, "pixels")
print("Channels    :", channels)
print("Data Type   :", img.dtype)
print("Resolution  :", width, "x", height)
# 6. Display Original Image

plt.figure(figsize=(7, 5))
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()

# Save original
cv2.imwrite("outputs/original.jpg", img)
# 7. Save Image as JPEG and PNG

cv2.imwrite("outputs/output_image.jpg", img)
cv2.imwrite("outputs/output_image.png", img)

print("\nJPEG and PNG images saved successfully.")
# 8. Grayscale Conversion
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2.imwrite("outputs/grayscale.jpg", gray)

plt.figure(figsize=(7, 5))
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()

# 9. HSV Color Space

hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Convert HSV to RGB only for displaying
hsv_display = cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)

cv2.imwrite("outputs/hsv.jpg", hsv_display)

plt.figure(figsize=(7, 5))
plt.imshow(hsv_display)
plt.title("HSV Image")
plt.axis("off")
plt.show()

# 10. LAB Color Space

lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)

# Convert LAB to RGB only for displaying
lab_display = cv2.cvtColor(lab, cv2.COLOR_LAB2RGB)

cv2.imwrite("outputs/lab.jpg", lab_display)

plt.figure(figsize=(7, 5))
plt.imshow(lab_display)
plt.title("LAB Image")
plt.axis("off")
plt.show()

# 11. Resize Image

resized = cv2.resize(img, (400, 300))

cv2.imwrite("outputs/resized.jpg", resized)

plt.figure(figsize=(7, 5))
plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.title("Resized Image (400 x 300)")
plt.axis("off")
plt.show()

# 12. Rotate Image

center = (width // 2, height // 2)

rotation_matrix = cv2.getRotationMatrix2D(
    center,
    45,
    1.0
)

rotated = cv2.warpAffine(
    img,
    rotation_matrix,
    (width, height)
)

cv2.imwrite("outputs/rotated.jpg", rotated)

plt.figure(figsize=(7, 5))
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.title("Rotated Image (45 Degrees)")
plt.axis("off")
plt.show()

# 13. Horizontal Flip

horizontal_flip = cv2.flip(img, 1)

cv2.imwrite(
    "outputs/horizontal_flip.jpg",
    horizontal_flip
)

plt.figure(figsize=(7, 5))
plt.imshow(cv2.cvtColor(horizontal_flip, cv2.COLOR_BGR2RGB))
plt.title("Horizontal Flip")
plt.axis("off")
plt.show()
# 14. Vertical Flip

vertical_flip = cv2.flip(img, 0)

cv2.imwrite(
    "outputs/vertical_flip.jpg",
    vertical_flip
)

plt.figure(figsize=(7, 5))
plt.imshow(cv2.cvtColor(vertical_flip, cv2.COLOR_BGR2RGB))
plt.title("Vertical Flip")
plt.axis("off")
plt.show()
# 15. Negative Image

negative = 255 - img

cv2.imwrite(
    "outputs/negative.jpg",
    negative
)

plt.figure(figsize=(7, 5))
plt.imshow(cv2.cvtColor(negative, cv2.COLOR_BGR2RGB))
plt.title("Negative Image")
plt.axis("off")
plt.show()
# 16. Region of Interest (ROI)

# ROI coordinates
x1 = 100
y1 = 100
x2 = min(400, width)
y2 = min(300, height)

roi = img[y1:y2, x1:x2]

if roi.size > 0:

    cv2.imwrite(
        "outputs/roi.jpg",
        roi
    )

    plt.figure(figsize=(7, 5))
    plt.imshow(cv2.cvtColor(roi, cv2.COLOR_BGR2RGB))
    plt.title("Region of Interest (ROI)")
    plt.axis("off")
    plt.show()

else:
    print("ROI coordinates are outside the image.")
# 17. Final Comparison

plt.figure(figsize=(15, 10))

# Original
plt.subplot(2, 3, 1)
plt.imshow(img_rgb)
plt.title("Original")
plt.axis("off")

# Grayscale
plt.subplot(2, 3, 2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

# Resized
plt.subplot(2, 3, 3)
plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.title("Resized")
plt.axis("off")

# Rotated
plt.subplot(2, 3, 4)
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.title("Rotated 45°")
plt.axis("off")

# Negative
plt.subplot(2, 3, 5)
plt.imshow(cv2.cvtColor(negative, cv2.COLOR_BGR2RGB))
plt.title("Negative")
plt.axis("off")

# ROI
plt.subplot(2, 3, 6)

if roi.size > 0:
    plt.imshow(cv2.cvtColor(roi, cv2.COLOR_BGR2RGB))
    plt.title("ROI")
else:
    plt.title("ROI Not Available")

plt.axis("off")

plt.tight_layout()

plt.savefig(
    "outputs/final_comparison.jpg",
    bbox_inches="tight"
)

plt.show()

print(" Experiment 1 Completed Successfully")

print("\nAll output images are saved in:")
print("outputs/")

print("\nFiles generated:")

for file in os.listdir("outputs"):
    print(" -", file)
```
